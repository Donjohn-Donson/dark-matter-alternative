```python
import sympy as sp
```

```python
rs, lam = sp.symbols("r_s lambda")   
```

```python
r =sp.Function("r")(lam)
t =sp.Function("t")(lam)
phi =sp.Function("phi")(lam)
r_d = r.diff(lam)
t_d = t.diff(lam)
phi_d = phi.diff(lam)
```

```python
gamma =  1-rs/r
gamma
```

```python
L = sp.sqrt(-gamma*t_d**2+gamma**(-1)*r_d**2+r**2*phi_d**2)
L
```

```python
Ls = sp.Function(r'L')(lam)
gammas = sp.Function(r'gamma')(lam)
p_r = L.diff(r_d).subs({L: Ls, gamma: gammas })
p_r
```

```python
Ls = sp.Function(r'L')(lam)
gammas = sp.Function(r'gamma')(lam)

sp.Eq(p_r.diff(lam),L.diff(r)).subs({L: Ls, gamma: gammas })
```
