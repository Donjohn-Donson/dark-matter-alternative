```python
from sympy import *

t, a, c = symbols('t a c', real = True)
z = Function('z')(t)
z_d = diff(z, t)
z_dd = diff(z, t, 2)
m = Function('m')(z)
m_dz = diff(m, z)
m_dt = diff(m, t)
```

Defining z's functional form and derivitaves:

```python
zs = 1/a*(sqrt((a*t)**2+1)-1)
zs
```

```python
zs_d = diff(zs, t)
zs_dd = diff(zs, t, 2)
```

Defining the Lagrangian in code:

```python
L = -m*sqrt(1-z_d**2)
L
```

Taking the partial of the Lagranian with repsect to z:

```python
diff(L, z)
```

Taking the derivative of L with respect to $\dot z$:

```python
diff(L, z_d)
```

Taking the derivative of $\frac{\partial L}{\partial \dot z}$ with respect to time:

```python
diff(diff(L, z_d), t)
```

creating EL EQ:

```python
EL = Eq(diff(L, z), diff(diff(L, z_d), t))
EL
```

Solving the Euler-Lagrange Eq for $\frac{\partial m}{\partial z}:$

```python
m_ode = Eq(m_dz,solve(EL,m_dz)[0])
m_ode
```

Pulling m to the left side & Splitting this into two integrals:

```python
m_ode.subs(m,1).rhs
```

Noting that:
$$
\int f(z(t))dz = \int f(z(t))\frac{dz}{dt}dt
$$

```python
ode_sol = Eq(ln(m),(integrate((Eq(m**-1*m_dz,solve(EL,m_dz)[0].subs(m,1)).rhs.subs(z,zs)*z_d).subs(z,zs).doit(), t))+c)
ode_sol
```

```python
solve(ode_sol,m)[0]
```

solving z for $\sqrt{a^{2} t^{2} + 1}$ gives:
$$
az+1  = \sqrt{a^{2} t^{2} + 1}
$$
this implies 
$$
m(t)=\frac{e^{c}}{\sqrt{a^{2} t^{2} + 1}}\rightarrow m(z)=\frac{e^{c}}{az+1}
$$

Plugging back into the Lagrangian:
$$
L = - \frac{e^{c}}{az+1} \sqrt{1 - \left(\frac{d}{d t} z{\left(t \right)}\right)^{2}} 
$$
