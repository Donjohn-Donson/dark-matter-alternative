```python
from sympy import *

t = symbols('t', real = True)
z = Function('z')(t)
z_d = diff(z, t)
z_dd = diff(z, t, 2)
m = Function('m')(z)
```

Defining the Lagrangian in code:

```python
L = -m*sqrt(1-z_d**2)
L
```

Taking the partial of the Lagranian with repsect to z:

```python
diff(L, z)
```

Taking the derivative of \frac{\partial L}{\partial z} with respect to time:

```python
diff(diff(L, z_d), t)
```
