```python
from sympy import *
from IPython.display import display, Math

r_sym = symbols('r', positive=True, real=True)
m_r = Function('m')(r_sym)
rdot, rddot = symbols(r'\dot{r} \ddot{r}', real=True)
L = symbols('L', positive=True, real=True)

ode = Eq(diff(m_r, r_sym) * (1 - rdot**2),
         m_r * rddot + L**2 / (m_r * r_sym**3))

display(Math(r"\text{ODE for } m(r): \quad " + latex(ode)))
```

```python
u = Function('u')(r_sym)
alpha = rddot / (1 - rdot**2)
beta  = L**2 / (1 - rdot**2)

# Bernoulli substitution u = m²  →  linear ODE
linear_ode = Eq(diff(u, r_sym) - 2*alpha*u, 2*beta / r_sym**3)
display(Math(r"\text{Linearised ODE for } u = m^2(r): \quad " + latex(linear_ode)))

sol_u = dsolve(linear_ode, u)
display(Math(r"u(r) = m^2(r): \quad " + latex(sol_u)))

m_sol = sqrt(sol_u.rhs)
display(Math(r"m(r) = " + latex(m_sol)))
```

```python
m_circ = Function('m')(r_sym)
ode_circ = Eq(diff(m_circ, r_sym), L**2 / (m_circ * r_sym**3))
sol_circ = dsolve(ode_circ, m_circ)
display(Math(r"\text{Circular orbit } (\dot{r}=\ddot{r}=0): \quad " + latex(ode_circ)))
display(Math(r"m(r) = " + latex(sol_circ.rhs)))

m_rad = Function('m')(r_sym)
ode_rad = Eq(diff(m_rad, r_sym) * (1 - rdot**2), m_rad * rddot)
sol_rad = dsolve(ode_rad, m_rad)
display(Math(r"\text{Radial free-fall } (L=0): \quad " + latex(ode_rad)))
display(Math(r"m(r) = " + latex(sol_rad.rhs)))
```

```python
import pygame
import math

# --- Parameters (tweak freely) ---
L   = 3.0    # angular momentum
C   = 20.0   # integration constant  (circular-orbit solution: m² = C - L²/r²)
r0  = 1.5    # orbital radius  — must satisfy C > L²/r0² for real mass

m0    = math.sqrt(C - L**2 / r0**2)   # m(r0) from circular-orbit solution
omega = L / (m0 * r0**2)              # θ̇ = L / (m r²)

# --- Window setup ---
W, H   = 700, 700
SCALE  = 120
CX, CY = W // 2, H // 2

pygame.init()
screen = pygame.display.set_mode((W, H))
pygame.display.set_caption("Circular orbit — variable mass")
clock  = pygame.time.Clock()
font   = pygame.font.SysFont("monospace", 16)

BLACK  = (10,  10,  10)
WHITE  = (255, 255, 255)
GREY   = (60,  60,  60)
BLUE   = (80,  160, 255)
YELLOW = (255, 220, 50)

theta = 0.0
dt    = 1 / 60

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    theta += omega * dt

    px = int(CX + r0 * math.cos(theta) * SCALE)
    py = int(CY - r0 * math.sin(theta) * SCALE)

    screen.fill(BLACK)
    pygame.draw.circle(screen, GREY,   (CX, CY), int(r0 * SCALE), 1)  # orbit ring
    pygame.draw.circle(screen, YELLOW, (CX, CY), 8)                    # source
    pygame.draw.circle(screen, BLUE,   (px, py), 10)                   # particle

    screen.blit(font.render(f"r0={r0:.2f}  L={L:.2f}  m(r0)={m0:.3f}  omega={omega:.3f}", True, WHITE), (12, 12))
    screen.blit(font.render(f"theta = {theta % (2*math.pi):.2f} rad", True, WHITE), (12, 32))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
