```python
import numpy as np
import matplotlib.pyplot as plt
import subprocess, os, platform

# Simple heuristic shape: rotation curve -> wavelength shift vs position
R_max = 15.0
r_scale = 3.0
v_max = 220.0
h_light = 4.0
c = 299792.458
lambda0 = 6562.8

x = np.linspace(-R_max, R_max, 500)
v_los = np.sign(x) * v_max * np.tanh(np.abs(x) / r_scale)
delta_lambda = lambda0 * v_los / c
fig, ax = plt.subplots(figsize=(6, 5))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

ax.plot(x, delta_lambda, color='black', lw=3, solid_capstyle='round')

ax.set_xlabel('Position across galaxy (⊥ to line of sight)')
ax.set_ylabel('Wavelength shift')
ax.xaxis.set_label_coords(0.5, -0.05)
ax.yaxis.set_label_coords(-0.08, 0.5)
ax.set_xticks([])

y_lim = lambda0 * v_max / c
ax.set_yticks([-y_lim, y_lim])
ax.set_yticklabels(['Red', 'Blue'])
for tick, color in zip(ax.get_yticklabels(), ['red', 'blue']):
    tick.set_color(color)

ax.set_title('Rotational Wavelength Shift')

ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
```

```python
fig.tight_layout()
filepath = r'C:\Users\Jonah\code\Research\plasma-quantum-group\dark-matter-alternative\Python files\Math\graphs\galaxy_rotation_simple.png'
fig.savefig(filepath, dpi=200, facecolor='white')
print("saved")

if platform.system() == 'Darwin':       # macOS
    subprocess.call(('open', filepath))
elif platform.system() == 'Windows':    # Windows
    os.startfile(filepath)
else:                                   # linux variants
    subprocess.call(('xdg-open', filepath))
```

```python
# Second graph: identical curve, shifted up and to the left relative to the central axis
shift_x = -2.5
shift_y = (lambda0 * v_max / c) * 0.2

v_los2 = np.sign(x - shift_x) * v_max * np.tanh(np.abs(x - shift_x) / r_scale)
delta_lambda2 = lambda0 * v_los2 / c + shift_y

fig2, ax2 = plt.subplots(figsize=(6, 5))
fig2.patch.set_facecolor('white')
ax2.set_facecolor('white')

ax2.plot(x, delta_lambda2, color='black', lw=3, solid_capstyle='round')

ax2.set_xlabel('Position across galaxy (⊥ to line of sight)')
ax2.set_ylabel('Wavelength shift')
ax2.xaxis.set_label_coords(0.5, -0.05)
ax2.yaxis.set_label_coords(-0.08, 0.5)
ax2.set_xticks([])

y_amp = lambda0 * v_max / c
ax2.set_yticks([shift_y - y_amp, shift_y + y_amp])
ax2.set_yticklabels(['Red', 'Blue'])
for tick, color in zip(ax2.get_yticklabels(), ['red', 'blue']):
    tick.set_color(color)

ax2.set_title('Rotational Wavelength Shift')

ax2.spines['left'].set_position(('data', 0))
ax2.spines['bottom'].set_position(('data', 0))
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
```
