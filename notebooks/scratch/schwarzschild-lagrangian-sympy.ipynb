# Schwarzschild Lagrangian in SymPy — $(-,+,+,+)$ signature

The geodesic Lagrangian is $\mathcal{L} = g_{\mu\nu}\dot{x}^\mu\dot{x}^\nu$ (dots = $d/d\lambda$).

In Schwarzschild coordinates $(t, r, \theta, \varphi)$ with natural units $G = c = 1$:

$$\mathcal{L} = -\left(1-\frac{2M}{r}\right)\dot{t}^2 + \left(1-\frac{2M}{r}\right)^{-1}\dot{r}^2 + r^2\dot{\theta}^2 + r^2\sin^2\!\theta\,\dot{\varphi}^2$$

```python
import sympy as sp
from IPython.display import display

sp.init_printing(use_latex='mathjax')   # renders via the notebook's MathJax engine

# ── Symbols ──────────────────────────────────────────────────────────────────
lam = sp.Symbol('lambda', real=True) 
tau = sp.Symbol('tau', real=True)         # affine parameter
M   = sp.Symbol('M', positive=True)           # Schwarzschild mass (G=c=1)

# Coordinates as functions of the affine parameter
t     = sp.Function('t'    )(lam)
r     = sp.Function('r'    )(lam)
theta = sp.Function('theta')(lam)
phi   = sp.Function('phi'  )(lam)
alpha   = sp.Function('alpha'  )(r)
Ls   = sp.Function('L'  )(lam)
lamt   = sp.Function('lambda'  )(tau)

coords = [t, r, theta, phi]

# Coordinate velocities
t_d, r_d, th_d, ph_d = [sp.diff(q, lam) for q in coords]

# ── Schwarzschild metric tensor g_μν (−,+,+,+) ───────────────────────────────
f = 1 - 2*M/r   # lapse factor

g = sp.diag(-f, 1/f, r**2, r**2 * sp.sin(theta)**2)

print("Metric tensor g_μν:")
display(g)
```

```python
# ── Lagrangian  L = g_μν ẋ^μ ẋ^ν  ────────────────────────────────────────────
vels = [t_d, r_d, ph_d]

L = sp.sqrt(sum(g[i, i] * vels[i]**2 for i in range(3)))

print("Lagrangian  L = g_μν ẋ^μ ẋ^ν :")
display(L.subs(1-2*M/r,alpha))

L = L.subs(1-2*M/r,alpha)
```

```python
# ── Euler-Lagrange equations  d/dλ(∂L/∂ẋ^μ) − ∂L/∂x^μ = 0 ──────────────────
def euler_lagrange(Lag, q):
    q_d = sp.diff(q, lam)
    EL  = sp.diff(sp.diff(Lag, q_d), lam) - sp.diff(Lag, q)
    return sp.simplify(EL)

labels = ['t', 'r', 'phi']
EL_eqs = {lab: euler_lagrange(L, q) for lab, q in zip(labels, coords)}

for lab, eq in EL_eqs.items():
    print(f"EL equation for {lab}:")
    display(sp.Eq(eq, 0).subs(L, Ls))
```

```python
# ── Conserved quantities (from cyclic coordinates) ────────────────────────────
# t and phi are cyclic  →  conjugate momenta are constants of motion

# Energy per unit mass  E = -∂L/∂ṫ  =  (1 - 2M/r) ṫ   [minus sign from (-,+,+,+)]
p_t   = sp.diff(L, t_d)
# Angular momentum per unit mass  ℓ = ∂L/∂φ̇  =  r² sin²θ φ̇
p_phi = sp.diff(L, ph_d)


display("p_t  = ∂L/∂ṫ  =", p_t)
display("p_φ  = ∂L/∂φ̇ =", p_phi)

# On the equatorial plane theta = pi/2
L_eq = L.subs(theta, sp.pi/2).subs(sp.diff(theta, lam), 0)
print("\nEquatorial Lagrangian (θ = π/2):")

L_eq.evalf()
```
