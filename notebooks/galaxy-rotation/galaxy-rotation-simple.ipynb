```python
import numpy as np
import matplotlib.pyplot as plt
import subprocess, os, platform
```

```python
# Simple heuristic shape: rotation curve -> wavelength shift vs position
R_max = 15.0
r_scale = 3.0
v_max = 220.0
h_light = 4.0
c = 299792.458
lambda0 = 6562.8

x = np.linspace(-R_max, R_max, 500)
v_los = np.sign(x) * v_max * np.tanh(np.abs(x) / r_scale)
delta_lambda = lambda0 * v_los / c
fig, ax = plt.subplots(figsize=(6, 5))
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

ax.plot(x, delta_lambda, color='black', lw=3, solid_capstyle='round')

ax.set_xlabel('Position across galaxy (⊥ to line of sight)')
ax.set_ylabel('Wavelength shift')
ax.xaxis.set_label_coords(0.5, -0.05)
ax.yaxis.set_label_coords(-0.08, 0.5)
ax.set_xticks([])

y_lim = lambda0 * v_max / c
ax.set_yticks([-y_lim, y_lim])
ax.set_yticklabels(['Red', 'Blue'])
for tick, color in zip(ax.get_yticklabels(), ['red', 'blue']):
    tick.set_color(color)

ax.set_title('Rotational Wavelength Shift')

ax.spines['left'].set_position(('data', 0))
ax.spines['bottom'].set_position(('data', 0))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
```

```python
fig.tight_layout()

from pathlib import Path
graphs_dir = Path.cwd().parent / "Math" / "graphs"
graphs_dir.mkdir(parents=True, exist_ok=True)
filepath = str(graphs_dir / "galaxy_rotation_simple.png")

fig.savefig(filepath, dpi=200, facecolor='white')

if platform.system() == 'Darwin':       # macOS
    subprocess.call(('open', filepath))
else:                                    # Windows
    os.startfile(filepath)
print("saved")
```

```python
# Second graph: identical curve, shifted to the left relative to the central axis
shift_x = -2.5

v_los2 = np.sign(x - shift_x) * v_max * np.tanh(np.abs(x - shift_x) / r_scale)
delta_lambda2 = lambda0 * v_los2 / c

fig2, ax2 = plt.subplots(figsize=(6, 5))
fig2.patch.set_facecolor('white')
ax2.set_facecolor('white')

ax2.plot(x, delta_lambda2, color='black', lw=3, solid_capstyle='round')

ax2.set_xlabel('Position across galaxy (⊥ to line of sight)')
ax2.set_ylabel('Wavelength shift')
ax2.xaxis.set_label_coords(0.5, -0.05)
ax2.yaxis.set_label_coords(-0.08, 0.5)
ax2.set_xticks([])

y_amp = lambda0 * v_max / c
ax2.set_yticks([-y_amp, y_amp])
ax2.set_yticklabels(['Red', 'Blue'])
for tick, color in zip(ax2.get_yticklabels(), ['red', 'blue']):
    tick.set_color(color)

ax2.set_title('Rotational Wavelength Shift')

ax2.spines['left'].set_position(('data', 0))
ax2.spines['bottom'].set_position(('data', 0))
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
```

```python
fig2.tight_layout()

from pathlib import Path
graphs_dir = Path.cwd().parent / "Math" / "graphs"
graphs_dir.mkdir(parents=True, exist_ok=True)
filepath2 = str(graphs_dir / "galaxy_rotation_simple_shifted.png")

fig2.savefig(filepath2, dpi=200, facecolor='white')

if platform.system() == 'Darwin':       # macOS
    subprocess.call(('open', filepath2))
    print(filepath2)
else:                                    # Windows
    os.startfile(filepath2)
    print(filepath2)
print("saved")
```

```python
# --- Heuristic "mystery force" curve — symmetric by construction ---
# Instead of shifting the signed curve f(x) by d and subtracting,
# work in unsigned radius r = |x|. Both terms below depend only on r,
# so D(x) = D(-x) exactly, for any value of `shift`.

shift = 2.5          # heuristic center offset, for illustration only
r = np.abs(x)        # unsigned galactocentric radius

f = lambda rr: v_max * np.tanh(rr / r_scale)   # same rotation-curve shape as before

delta_lambda3 = lambda0 * (f(r + shift) - f(r)) / c

fig3, ax3 = plt.subplots(figsize=(6, 5))
fig3.patch.set_facecolor('white')
ax3.set_facecolor('white')

ax3.plot(x, delta_lambda3, color='black', lw=3, solid_capstyle='round')

ax3.set_xlabel('Position across galaxy (⊥ to line of sight)')
ax3.set_ylabel('Wavelength shift difference')
ax3.xaxis.set_label_coords(0.5, -0.05)
ax3.yaxis.set_label_coords(-0.08, 0.5)
ax3.set_xticks([])
ax3.set_yticks([])

ax3.set_title('Difference Graph')

ax3.spines['left'].set_position(('data', 0))
ax3.spines['bottom'].set_position(('data', 0))
ax3.spines['top'].set_visible(False)
ax3.spines['right'].set_visible(False)
```

```python
fig3.tight_layout()

from pathlib import Path
graphs_dir = Path.cwd().parent / "Math" / "graphs"
graphs_dir.mkdir(parents=True, exist_ok=True)
filepath3 = str(graphs_dir / "galaxy_rotation_simple_difference.png")

fig3.savefig(filepath3, dpi=200, facecolor='white')

if platform.system() == 'Darwin':       # macOS
    subprocess.call(('open', filepath3))
    print(filepath3)
else:                                    # Windows
    os.startfile(filepath3)
    print(filepath3)
print("saved")
```

```python
# Fourth graph: m(r) = sqrt(c + 2E^2/r - L^2/r^2)
# Illustrative constants (adjust E, L, c_const to your actual values)
# NOTE: named c_const here (not c) to avoid clobbering the speed of light `c` defined above
E = 2
L = 2
c_const = 1.0

# sqrt argument f(r) = c_const + 2E^2/r - L^2/r^2 has a single interior maximum
# at r* = L^2/E^2, is -inf as r->0+, and -> c_const as r->inf.
# For c_const >= 0 the valid domain is [r_min, inf); r_min is the positive root of
# c_const*r^2 + 2E^2*r - L^2 = 0.
if c_const != 0:
    r_min = (-E**2 + np.sqrt(E**4 + c_const * L**2)) / c_const
else:
    r_min = L**2 / (2 * E**2)  # linear case (c_const = 0)

r_max = 10 * r_min
r = np.linspace(r_min, r_max, 500)

m_arg = c_const + 2 * E**2 / r - L**2 / r**2
m = np.sqrt(np.clip(m_arg, 0, None))

fig4, ax4 = plt.subplots(figsize=(6, 3))
fig4.patch.set_facecolor('white')
ax4.set_facecolor('white')

ax4.plot(r, m, color='black', lw=3, solid_capstyle='round')

ax4.set_xlabel('r')
ax4.set_ylabel('m(r)')
ax4.xaxis.set_label_coords(0.5, -0.05)
ax4.yaxis.set_label_coords(-0.08, 0.5)
ax4.set_xlim(0, r_max)
ax4.set_ylim(bottom=0)
ax4.set_xticks([])
ax4.set_yticks([])



ax4.spines['left'].set_position(('data', 0))
ax4.spines['bottom'].set_position(('data', 0))
ax4.spines['top'].set_visible(False)
ax4.spines['right'].set_visible(False)
```
