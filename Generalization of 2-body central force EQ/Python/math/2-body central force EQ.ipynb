```python
from sympy import *
```

```python
t = symbols(r't')
phi = Function(r'\phi')(t)
r = Function('r')(t)
U = Function('U')(r)
M = Function('M')(r)
r_d = r.diff(t)
r_dd = r.diff(t).diff(t)

phi_d = phi.diff(t)
phi_dd = phi.diff(t).diff(t)
```

```python
L = 1/2*M*(r_d**2+(r*phi_d)**2) - U
L
```

```python
L.diff(r).simplify()
```

```python
L.diff(r).simplify()
```

```python
L.diff(r_d).diff(t).simplify()
```

```python
solve(Eq(L.diff(r_d).diff(t).simplify(),L.diff(r).simplify()),r_dd)[0].simplify()
```
